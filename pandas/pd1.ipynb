# Series

```python
import pandas as pd
```

```python
list_s = [1,2,-3,6.2,'data values']
print(list_s)
```

```python
series1 = pd.Series(list_s)
print(series1)
```

```python
type(series1)
```

```python
empty_s = pd.Series([])
print(empty_s)
```

```python
series2 = pd.Series([1,2,3,4],index=['a','b','c','d'])
print(series2)
```

```python
series2 = pd.Series([1,2,3,4],index=['a','b','c','d'],dtype=float)
print(series2)
```

```python
series2 = pd.Series([1,2,3,4],index=['a','b','c','d'],dtype=float,name='data values')
print(series2)
```

```python
scaler_s = pd.Series(0.5)
print(scaler_s)
```

```python
scaler_s = pd.Series(0.5,index=[1,2,3])
print(scaler_s)
```

```python
dict_s = pd.Series({'a':1,'b':2})
print(dict_s)
```
