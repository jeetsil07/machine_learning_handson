# Data cleaning
### Missing value imputation using scikit-learn
### for numeric and categorical data

```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
```

```python
# load data set
train_path = r'E:\machine_learning_practice\practice\machine_learning_handson\data\train.csv'
test_path = r'E:\machine_learning_practice\practice\machine_learning_handson\data\test.csv'
train = pd.read_csv(train_path)
test = pd.read_csv(test_path)
```

```python
train.shape
```

```python
test.shape
```

```python
X_train = train.drop(columns='SalePrice')
y_train= train['SalePrice']
print("shape of X_train",X_train.shape)
print("shape of y_train",y_train.shape)
```

# numerical missing value imputation

```python
num_vars = X_train.select_dtypes(include = ["int64","float64"]).columns
```

```python
num_vars
```

```python
X_train[num_vars].isnull().sum()
```

```python
imputer_mean = SimpleImputer(strategy='mean')
# imputer_mean = SimpleImputer(strategy='constant',fill_value=99) add constant
```

```python
imputer_mean.fit(X_train[num_vars])
```

```python
imputer_mean.statistics_
```

```python
imputer_mean.transform(X_train[num_vars])
```

```python
X_train[num_vars] = imputer_mean.transform(X_train[num_vars])
test[num_vars] = imputer_mean.transform(test[num_vars])
```

```python
X_train[num_vars].isnull().sum()
```

```python
test[num_vars].isnull().sum()
```

# categorical missing value imputation

```python
cat_vars = X_train.select_dtypes(include = ["object"]).columns
cat_vars
```

```python
X_train[cat_vars].isnull().sum()
```

```python
imputer_mode = SimpleImputer(strategy='most_frequent')
# imputer_mode = SimpleImputer(strategy='constant',fill_value='str') add constant
```

```python
imputer_mode.fit(X_train[cat_vars])
```

```python
X_train[cat_vars] = imputer_mode.transform(X_train[cat_vars])
test[cat_vars] = imputer_mode.transform(test[cat_vars])
```

```python
X_train[cat_vars].isnull().sum()
```

```python
test[cat_vars].isnull().sum()
```
