```python
missing_val_perc =  df.isnull().sum()/df.shape[0] *100
missing_val_perc
```

```python
missing_val_clm_gre_20 = missing_val_perc[missing_val_perc > 20].keys()
missing_val_clm_gre_20
```

```python
df2_drp_clm = df.drop(columns = missing_val_clm_gre_20)
df2_drp_clm.shape
```

```python
df3_num=df2_drp_clm.select_dtypes(include = ['int64','float64'])
df3_num
```

```python
plt.figure(figsize=(16,9))
sns.heatmap(df3_num.isnull())
```

```python
df3_num[df3_num.isnull().any(axis=1)]
```

```python
df3_num.isnull().sum()
```

```python
missing_num_var = [var for var in df3_num.columns if df3_num[var].isnull().sum()>0]
missing_num_var
```

```python
plt.figure(figsize=(10,10))
sns.set()
for i, var in enumerate(missing_num_var):
    plt.subplot(2,2,i+1)
    # sns.histplot(df[var],bins = 20, kde=True)
    sns.histplot(df3_num[var], bins= 20, kde=True)

plt.tight_layout()
plt.show()
```

```python
df4_num_mean = df3_num.fillna(df3_num.mean())
df4_num_mean.isnull().sum().sum()
```

```python
plt.figure(figsize=(10,10))
sns.set()
for i, var in enumerate(missing_num_var):
    plt.subplot(2,2,i+1)
    sns.histplot(df4_num_mean[var],bins = 20, kde=True)
    sns.histplot(df3_num[var], bins= 20, kde=True)

plt.tight_layout()
plt.show()
```

```python
df5_num_median = df3_num.fillna(df3_num.median())
df5_num_median.isnull().sum().sum()
```

```python
plt.figure(figsize=(10,10))
sns.set()
for i, var in enumerate(missing_num_var):
    plt.subplot(2,2,i+1)
    sns.histplot(df4_num_mean[var],bins = 20, kde=True)
    sns.histplot(df5_num_median[var],bins = 20, kde=True)
    sns.histplot(df3_num[var], bins= 20, kde=True)

plt.tight_layout()
plt.legend(labels=['original','mean','median'])
plt.show()
```

```python
for i, var in enumerate(missing_num_var):
    plt.figure(figsize=(10,10))
    plt.subplot(3,1,1)
    sns.boxplot(x=df[var])
    plt.subplot(3,1,2)
    sns.boxplot(x=df4_num_mean[var])
    plt.subplot(3,1,3)
    sns.boxplot(x=df5_num_median[var])
```

```python
df_concat = pd.concat([df3_num[missing_num_var],df4_num_mean[missing_num_var],df5_num_median[missing_num_var]],axis=1)
```
