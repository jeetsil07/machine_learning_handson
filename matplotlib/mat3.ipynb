```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib import style
```

# bar plot

```python
classes = ['Python','R','AI','ML','DS']
class1_students = [30,10,20,25,10]
class2_students = [40,5,20,20,10]
class3_students = [35,5,30,15,15]
```

```python
plt.bar(classes,class1_students)
plt.show()
```

```python
plt.barh(classes,class1_students)
plt.show()
```

```python
plt.bar(classes,class1_students,width = 0.2,align = 'edge', color='y', edgecolor='m',
       linewidth=5,alpha=0.6, linestyle='--',label="class1 students")
plt.show()
```

```python
style.use('ggplot')
plt.figure(figsize=(16,9))
plt.bar(classes,class1_students,width = 0.6,align = 'edge', color='k', edgecolor='m',
       linewidth=5,alpha=0.6, linestyle='--',label="class1 students")
plt.show()
```

```python
style.use('ggplot')
plt.figure(figsize=(16,9))
plt.bar(classes,class1_students,align = 'edge', color='k', edgecolor='m',
       linewidth=5,alpha=0.6, linestyle='--',label="class1 students")
plt.title('Bar chart of class', fontsize=25)
plt.xlabel('classes',fontsize=15)
plt.ylabel('no of students',fontsize=15)
plt.show()
```

```python
style.use('ggplot')
plt.figure(figsize=(16,9))
plt.bar(classes,class1_students,color='b',label="class1 students")
plt.bar(classes,class2_students,color='g',label="class2 students")
plt.bar(classes,class3_students,color='k',label="class3 students")
plt.title('Bar chart of class', fontsize=25)
plt.xlabel('classes',fontsize=15)
plt.ylabel('no of students',fontsize=15)
plt.show()
```

```python
style.use('ggplot')
plt.figure(figsize=(16,9))
classes_index = np.arange(len(classes))
print(classes_index)
width = 0.2
plt.bar(classes_index,class1_students,color='b',width=width,label="class 1 students")
plt.bar(classes_index+width,class2_students,color='g',width=width,label="class 2 students")
plt.bar(classes_index+2*width,class3_students,color='r',width=width,label="class 3 students")
plt.title('Bar chart of class', fontsize=25)
plt.xlabel('classes',fontsize=15)
plt.ylabel('no of students',fontsize=15)
plt.xticks(classes_index + width, classes)
plt.show()
```

```python
style.use('ggplot')
plt.figure(figsize=(16,9))
classes_index = np.arange(len(classes))
print(classes_index)
height = 0.2
plt.barh(classes_index,class1_students,color='b',height=height,label="class 1 students")
plt.barh(classes_index+height,class2_students,color='g',height=height,label="class 2 students")
plt.barh(classes_index+2*height,class3_students,color='r',height=height,label="class 3 students")
plt.title('Bar chart of class', fontsize=25)
plt.xlabel('classes',fontsize=15)
plt.ylabel('no of students',fontsize=15)
plt.yticks(classes_index + height, classes)
plt.show()
```
