```python
import matplotlib.pyplot as plt
import pandas as pd
```

# Pie chart

```python
classes = ["Python","R","Machine Learning","Artificial Intelligence","Data Science"]
class1_students= [45,15,35,25,30]

plt.pie(class1_students,labels=classes)
plt.show()
```

```python
classes = ["Python","R","Machine Learning","Artificial Intelligence","Data Science"]
class1_students= [45,15,35,25,30]
colors=["c","b","r","y","g"]
explode= [0.03,0,0.1,0,0]
textprops = {"fontsize":15}
plt.pie(class1_students,labels=classes,explode=explode, colors=colors,
        autopct="%0.2f%%",shadow = True, radius= 1.4, startangle = 90,
        textprops=textprops
       )
plt.show()
```

```python
plt.figure(figsize=(8,4))
classes = ["Python","R","Machine Learning","Artificial Intelligence","Data Science"]
class1_students= [45,15,35,25,30]
colors=["c","b","r","y","g"]
explode= [0.03,0,0.1,0,0]
textprops = {"fontsize":15}
wedgeprops = {"linewidth":3,"width":3,"edgecolor":'k'}
plt.pie(class1_students,labels=classes,explode=explode, colors=colors,
        autopct="%0.2f%%",shadow = True, radius= 1.2, startangle = 270,
        textprops=textprops, pctdistance = 0.7, labeldistance = 1.6,
        counterclock = True, wedgeprops = wedgeprops, center=(2,3), 
        # frame = True,
        # rotatelabels = True
       )
plt.show()
```
