# Scatter plot

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
data_path = r"E:\machine_learning_practice\practice\machine_learning_handson\data\googleplaystore.csv"
```

```python
df_google_play_store_apps = pd.read_csv(data_path)
df_google_play_store_apps
```

```python
df_google_play_store_apps = pd.read_csv(data_path,nrows=1000)
df_google_play_store_apps
```

```python
x = df_google_play_store_apps['Rating']
y = df_google_play_store_apps['Reviews']
```

```python
plt.scatter(x,y)
plt.xlabel('Ratings')
plt.ylabel('Reviews')
plt.title('Google play store app scatter')
plt.show()
```

```python
plt.figure(figsize=(16,9))
plt.scatter(x,y, c = "r", marker='*',s = 100,alpha=0.2, linewidth=10, edgecolor='g')
plt.xlabel('Ratings')
plt.ylabel('Reviews')
plt.title('Google play store app scatter')
plt.show()
```

```python
plt.figure(figsize=(16,9))
plt.scatter(x,y, c = "r", marker='*',s = 100,alpha=0.2, linewidth=10, edgecolor='g')
plt.scatter(x,df_google_play_store_apps['Installs'], c = "y", marker='o',s = 100,alpha=0.2, linewidth=10, edgecolor='r')
plt.xlabel('Ratings')
plt.ylabel('Reviews & Installs')
plt.title('Google play store app scatter')
plt.show()
```
