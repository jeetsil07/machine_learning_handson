```python
datapath = r'E:\machine_learning_practice\practice\machine_learning_handson\data\Who_is_responsible_for_global_warming.csv'
global_warming_df = pd.read_csv(datapath)
global_warming_df
```

```python
global_warming_df=global_warming_df.drop(columns = ['Country Code','Indicator Name','Indicator Code'],axis = 1).set_index('Country Name')
global_warming_df.head()
```

```python
plt.figure(figsize=(16,9))
sns.heatmap(global_warming_df)
```

```python
sns.heatmap(global_warming_df,cmap='coolwarm')
```

```python
plt.figure(figsize=(16,9))
sns.heatmap(global_warming_df,cmap='coolwarm',annot=True)
```

```python
plt.figure(figsize=(16,9))
annot_kws = {
    'fontsize': 10,
    'color':'k',
    'alpha': 0.6,
    'rotation':'vertical',
    'verticalalignment':'center',
    'backgroundcolor':'w'
}
sns.heatmap(global_warming_df,cmap='coolwarm',annot=True, annot_kws=annot_kws, linewidth=4, linecolor='k')
```

```python
global_warming_df.corr()
```

```python
plt.figure(figsize=(16,9))
ax=sns.heatmap(global_warming_df.corr(),cmap='coolwarm',annot=True,linewidth=3)
ax.tick_params(size=10,color='w',labelcolor='w',labelsize=10)
plt.title('who is responsible for global warming')
```

```python
from sklearn.datasets import load_breast_cancer
cancer_dataset = load_breast_cancer()
cancer_dataset
```

```python
cancer_df = pd.DataFrame(np.c_[cancer_dataset['data'],cancer_dataset['target']],
                        columns= np.append(cancer_dataset['feature_names'],['target']))
cancer_df
```

```python
plt.figure(figsize=(16,9))
ax=sns.heatmap(cancer_df.corr(),annot=True,linewidth=3)
ax.tick_params(size=10,color='w',labelcolor='r',labelsize=10)
plt.title('heatmap of breast cancer patient',fontsize=25)
plt.show()
```
