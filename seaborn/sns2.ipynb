# Histogram/Displot

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
tips_df = sns.load_dataset('tips')
tips_df
```

```python
sns.displot(tips_df['size'], kde=True)
```

```python
sns.displot(tips_df['tip'], kde=True)
```

```python
sns.displot(tips_df['total_bill'], kde=True)
```

```python
bins = [1,5,10,15,20,25,30,35,40,45,50,55]
plt.figure(figsize=(16,9))
sns.set()
sns.displot(tips_df['total_bill'],bins=bins)
plt.title("Histogram of Total Bill")
plt.xticks(bins)
plt.show()
```
