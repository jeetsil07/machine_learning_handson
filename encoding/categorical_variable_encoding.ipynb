# one hot encoding with sklearn

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
oh_enc = OneHotEncoder(sparse_output=False,drop='first')
```

```python
oh_enc_arr = oh_enc.fit_transform(tips_df[['sex', 'smoker', 'day', 'time']])
oh_enc_arr
```

```python
oh_enc_df = pd.DataFrame(oh_enc_arr, columns=oh_enc.get_feature_names_out(['sex', 'smoker', 'day', 'time']))
oh_enc_df
```

```python
final_df = pd.concat([tips_df[['total_bill', 'tip', 'size']], oh_enc_df], axis=1)
final_df
```
