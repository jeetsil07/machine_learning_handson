# Python numpy array slicing

```python
import numpy as np
```

```python
mx = np.arange(1,101).reshape(10,10)
print(mx)
```

```python
mx[0,0]
```

```python
mx[0,0].ndim
```

```python
mx[:,0]
```

```python
mx[:,0:1]
```

```python
mx[0,:]
```

```python
mx[0:1,:]
```

```python
mx[0:3,:]
```

```python
mx[1:4,1:4]
```

```python
mx.item
```
