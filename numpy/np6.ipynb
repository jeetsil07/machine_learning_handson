# Python numpy array concatenation and split

```python
import numpy as np
```

```python
arr1 = np.arange(1,17).reshape(4,4)
print(arr1)
```

```python
arr2 = np.arange(17,33).reshape(4,4)
print(arr2)
```

```python
list1 = [1,2,3,4]
list2 = [5,6,7,8]
list1+list2
```

```python
arr1+arr2
```

```python
np.concatenate((arr1,arr2))
```

```python
np.concatenate((arr1,arr2),axis = 1) # row wise
```

```python
np.concatenate((arr1,arr2),axis = 0) # column wise
```

```python
np.vstack((arr1,arr2))
```

```python
np.hstack((arr1,arr2))
```

```python
np.split(arr1,2)
```

```python
list1 = np.split(arr1,2)
type(list1)
```

```python
list1[0]
```

```python
np.split(arr1,2,axis = 1)
```
