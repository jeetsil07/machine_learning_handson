```python
import numpy as np
```

# arrange()

```python
arr_1d = np.arange(1,13)
print(arr_1d)
```

```python
even_arr = np.arange(1,13,2)
print(even_arr)
```

# linspace()

```python
np.linspace(1,5,6)
```

# reshape()

```python
arr_2d = arr_1d.reshape(2,6)
arr_2d
```

```python
arr_3d = arr_1d.reshape(2,3,2)
arr_3d
```

# ravel()

```python
arr_3d.ravel()
```

# flatten()

```python
arr_3d.flatten()
```

# transpose()

```python
arr_2d.transpose()
```
